```julia
using JLD2,Plots
```

```julia
st=load("data.jld2")
```

```julia
num_spin=2
num_valley=2
sys_size=36
```

```julia
plot_eigenvalues=zeros(Float64,num_spin,num_valley,sys_size,2)
for ja in 1:num_spin, jb in 1:num_valley, jc in 1:sys_size
    plot_eigenvalues[ja,jb,jc,:]=real.(st["HF_eigenvalue"][ja,jb,jc])
end
```

```julia
count=0
for ja in eachindex(vec(plot_eigenvalues))
  if vec(plot_eigenvalues)[ja]<1606.3719903056226
    count+=1
  end
end
```
