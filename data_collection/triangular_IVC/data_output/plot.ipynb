```julia
using JLD2, Plots
```

```julia
st=load("trywithpining.jld2")
```

```julia
a1m=[1/2,√3/2]
a2m=[1,0]
xgrid=zeros(Float64,50,50)
ygrid=zeros(Float64,50,50)    

for ja in 1:50, jb in 1:50
    rvec=ja/25*a1m+jb/25*a2m
 xgrid[ja,jb]=rvec[1]
 ygrid[ja,jb]=rvec[2]
  
end
```

```julia

plot(xgrid,ygrid,real.(st["HFdensity"][1]),st=:surf,axis_ratio=:equal,camera=[0,90])
xlims!(0,3)
ylims!(0,3)
```

```julia

plot(xgrid,ygrid,real.(st["HFdensity"][2]),st=:surf,axis_ratio=:equal,camera=[0,90])
xlims!(0,3)
ylims!(0,3)
```

```julia
Nq=3
dimension=37*2
HF_pathvalue=zeros(Float64,dimension,Nq^2)
for ja in 1:Nq^2
    HF_pathvalue[:,ja]=real(st["HFeigenvalue"][ja])
end 
```

```julia
myplot=plot()
plot!(1:Nq^2,Float64.(HF_pathvalue[1:16,:]'),legend=false,color="blue")
display(myplot)
```
