```julia
using JLD2

ϕspace=[0.0] #convert this to radian 
vfspace=collect(0.5:0.2:4.5)
V0=collect(1.0:0.2:6.4)
spin=[1.0,2.0]
bandspace=[2.0,3.0]
index=Vector{Float64}[]
count=0
for ja in spin, jb in V0, jc in vfspace, jd in ϕspace, je in bandspace
    if !isfile("search7/$(jc)vf$(ja)spin$(jb)V0$(jd)phi$(je)band.jld2")
        push!(index,[ja,jb,jc,jd,je])
    end

    if isfile("search7/$(jc)vf$(ja)spin$(jb)V0$(jd)phi$(je)band.jld2")
        
        count+=1
    end
end
```

```julia
index
jldsave("missedindex.jld2",missedindex=index)
```
