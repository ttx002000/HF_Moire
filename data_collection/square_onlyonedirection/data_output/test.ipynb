```julia
using JLD2, Plots
st=load("3.0Nq2.0flux2.0Vx0.0Vy1.0scale1.5constq1.0try.jld2")
```

```julia
Nq=3
dimension=49
HF_pathvalue=zeros(Float64,dimension,Nq^2)
for ja in 1:Nq^2
    HF_pathvalue[:,ja]=real(st["single_eigenvalue"][ja])
end 
```

```julia
myplot=plot()
plot!(1:Nq^2,Float64.(HF_pathvalue[1:10,:]'),legend=false,color="blue")
display(myplot)
```
