```julia
using JLD2
using Plots
```

```julia
st=load("try.jld2")
```

```julia

dimension=3*61
Nq=3
NOHF_pathvalue=zeros(Float64,dimension,Nq^2,2)
HF_pathvalue=zeros(Float64,dimension,Nq^2,2)
for ja in 1:Nq^2,vi in 1:2
    NOHF_pathvalue[:,ja,vi]=real((st["single_eigenvalue"][ja][vi]))
    HF_pathvalue[:,ja,vi]=real(st["HF_eigenvalue"][ja][vi])
end 
```

```julia
Parnum=2*Nq^2*(2*61)-2*Nq^2
```

```julia
a1m=[√3/2,1/2]
a2m=[0,1]
N3=50
xgrid=zeros(Float64,N3,N3)
ygrid=zeros(Float64,N3,N3)



for ja in 1:N3, jb in 1:N3
  xgrid[ja,jb]=(ja/N3*a1m[1]+jb/N3*a2m[1])
  ygrid[ja,jb]=(ja/N3*a1m[2]+jb/N3*a2m[2])

end
```

```julia
plot(xgrid,ygrid,abs.(real(st["HFdensity"][:,:,2,1])),st=:surf,axis_ratio=:equal,camera=[0,90])
xlims!(0,1.5)
ylims!(0,1.5)
```
